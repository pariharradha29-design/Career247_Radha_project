```python
#Grouped stats: average MonthlySpend by State, Education, Gender

df.groupby(["State","Education","Gender"])["MonthlySpend"].mean().reset_index()
```

```python
#Independent t-test

from scipy import stats

male_spend = df[df['Gender'] == 'Male']['MonthlySpend']
female_spend = df[df['Gender'] == 'Female']['MonthlySpend']
t_stat, p_value = stats.ttest_ind(male_spend, female_spend)
print("T-Statistic:", t_stat)
print("P-Value:", p_value)


if p_value < 0.05:
    print("Reject H0")
    print("There is a significant difference in spending between males and females.")
else:
    print("Fail to Reject H0")
    print("There is no significant difference in spending between males and females.")
```
