```python
df1["JoinDate"]=pd.to_datetime(df1["JoinDate"],errors="coerce")
df1
```

```python
df=df2["CustomerID"].isin(df1["CustomerID"])
df
```

```python
df=pd.merge(df1,df2,on="CustomerID",how="inner")
df
```

```python
rec=df2["TransactionDate"].max()+pd.Timedelta(days=1)
rec
```

```python
from datetime import date
```

```python
today = pd.Timestamp(date.today())

df_rfm = df.groupby("CustomerID").agg(Recency=("TransactionDate", lambda x: (today - x.max()).days),
    Frequency=("TransactionDate", lambda x: x.count()),
    Monetary=("TransactionAmount", lambda x: x.sum()))
df_rfm
```

```python
df_rfm["R_score"]=pd.cut(df_rfm["Recency"].rank(method="first"),5,labels=[5,4,3,2,1])
df_rfm["F_score"]=pd.cut(df_rfm["Frequency"].rank(method="first"),5,labels=[1,2,3,4,5])
df_rfm["M_score"]=pd.cut(df_rfm["Monetary"].rank(method="first"),5,labels=[1,2,3,4,5])
df_rfm
```

```python
df_rfm
```

```python
df_rfm["RFM_Segment"]=df_rfm["R_score"].astype(str)+df_rfm["F_score"].astype(str)+df_rfm["M_score"].astype(str)
df_rfm
```

```python
import numpy as np
```

```python
conditions = [
    # Champions: R 4-5, F 4-5, M 4-5
    (df_rfm["RFM_Segment"].str[0].astype(int).isin([4,5])) & 
    (df_rfm["RFM_Segment"].str[1].astype(int).isin([4,5])) & 
    (df_rfm["RFM_Segment"].str[2].astype(int).isin([4,5])),

    # Loyal Customers: F 4-5, R 2-5
    (df_rfm["RFM_Segment"].str[1].astype(int).isin([4,5])) & 
    (df_rfm["RFM_Segment"].str[0].astype(int).isin([2,3,4,5])),

    # Potential Loyalist: R 4-5, F 2-3
    (df_rfm["RFM_Segment"].str[0].astype(int).isin([4,5])) & 
    (df_rfm["RFM_Segment"].str[1].astype(int).isin([2,3])),

    # At Risk: R 1-2, F 3-5
    (df_rfm["RFM_Segment"].str[0].astype(int).isin([1,2])) & 
    (df_rfm["RFM_Segment"].str[1].astype(int).isin([3,4,5])),

    # Lost: R 1, F 1-2, M 1-2
    (df_rfm["RFM_Segment"].str[0].astype(int) == 1) & 
    (df_rfm["RFM_Segment"].str[1].astype(int).isin([1,2])) & 
    (df_rfm["RFM_Segment"].str[2].astype(int).isin([1,2])),

    # Big Spenders: M 4-5, F 2-3, R 3-4
    (df_rfm["RFM_Segment"].str[2].astype(int).isin([4,5])) & 
    (df_rfm["RFM_Segment"].str[1].astype(int).isin([2,3])) & 
    (df_rfm["RFM_Segment"].str[0].astype(int).isin([3,4])),]

segment_labels = ["Champions","Loyal Customers","Potential Loyalist","At Risk","Lost","Big Spenders"]

df_rfm["RFM_Segment_Name"] = np.select(conditions, segment_labels, default="Others")

df_rfm
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Count of customers in each segment 
plt.figure(figsize=(5,5))

sns.countplot(data=df_rfm,x="RFM_Segment_Name",hue="RFM_Segment_Name",legend=False,palette="viridis")

plt.xticks(rotation=45,ha='right')
plt.title("Customer Distribution by RFM Segment")
plt.xlabel("RFM Segment")
plt.ylabel("Customer Count")

plt.tight_layout()
plt.show()

 # Insights:- Customer distribution is concentrated in mid-value segments (Loyal and Potential Loyalists),
 # while high-value customers (Champions and Big Spenders) are relatively low,
 # indicating strong retention potential but an opportunity to improve revenue by upgrading customers to higher-value segments.
 
```

```python
#Revenue contribution per segment
rev = df_rfm.groupby("RFM_Segment_Name")["Monetary"].sum().reset_index()

plt.figure(figsize=(4,4))
sns.barplot(data=rev, x="RFM_Segment_Name", y="Monetary",hue="RFM_Segment_Name",legend=False,palette="Set2")

plt.xticks(rotation=45, ha='right')
plt.title("Revenue Contribution by RFM Segment")
plt.xlabel("RFM Segment")
plt.ylabel("Total Revenue")

plt.tight_layout()
plt.show()

#Insights:-Revenue is primarily driven by Loyal Customers and the “Others” segment,
#while Big Spenders contribute the least,
#indicating a reliance on mid-value customers and an opportunity to grow high-value customer revenue.
```

```python
#Recency vs Monetary scatter plot colored by segment

plt.figure(figsize=(10,4))
plt.title("Recency vs Monetary by RFM Segment")
sns.scatterplot(data=df_rfm,x="Recency",y="Monetary",hue="RFM_Segment_Name",palette="Set2",style="RFM_Segment_Name",alpha=0.7)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Insights:-High-value customers (Champions and Loyal Customers) cluster at low recency with high monetary value,
# while At Risk and Lost customers show higher recency and lower spending,
# indicating clear opportunities for retention and re-engagement strategies.
```

```python
segment_rev = df_rfm.groupby("RFM_Segment_Name")["Monetary"].sum().reset_index()

segment_rev = segment_rev.sort_values(by="Monetary", ascending=False).reset_index(drop=True)

segment_rev["Cumulative_%"] = segment_rev["Monetary"].cumsum() / segment_rev["Monetary"].sum()

fig, ax1 = plt.subplots(figsize=(6,5))
ax1.bar(segment_rev["RFM_Segment_Name"], segment_rev["Monetary"])
ax1.set_ylabel("Revenue")
ax1.tick_params(axis='x', rotation=45)
ax2 = ax1.twinx()
ax2.plot(segment_rev["RFM_Segment_Name"], segment_rev["Cumulative_%"],color='red', marker='o')
ax2.axhline(0.8,color='green', linestyle='--')
ax2.set_ylabel("Cumulative %")

plt.title("Segment Pareto Analysis")
plt.tight_layout()
plt.show()

# Insights:-Loyal Customers, Others, and Champions—contribute to the majority of revenue (around 80%),
# indicating a strong segment-level Pareto effect and highlighting where the business should focus retention and growth efforts.
```
